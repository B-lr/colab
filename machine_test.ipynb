Feature Engineering

```python
# Ensure Date is in datetime format
temp1['Date'] = pd.to_datetime(temp1['Date'])

# Remove leap day (Feb 29)
temp1 = temp1[~((temp1['Date'].dt.month == 2) & (temp1['Date'].dt.day == 29))]

# Separate data: 2005-2014 and 2015
df1 = temp1[temp1['Date'].dt.year != 2015]  # Past years
df2015 = temp1[temp1['Date'].dt.year == 2015]  # Year 2015
```

Computing record temperature

```python
# Compute min/max for past years
record_high = df1.groupby(df1['Date'].dt.dayofyear)["Data_Value"].max()
record_low = df1.groupby(df1['Date'].dt.dayofyear)["Data_Value"].min()

# Compute min/max for 2015
record_high_2015 = df2015.groupby(df2015['Date'].dt.dayofyear)["Data_Value"].max()
record_low_2015 = df2015.groupby(df2015['Date'].dt.dayofyear)["Data_Value"].min()

record_high_2015 = record_high_2015.reindex(record_high.index)
record_low_2015 = record_low_2015.reindex(record_low.index)


# Identify record-breaking points
broken_highs = record_high_2015[record_high_2015 > record_high].index
broken_lows = record_low_2015[record_low_2015 < record_low].index
```

Plotting the graph

```python
import matplotlib.pyplot as plt


# Plotting
plt.figure(figsize=(20,7))

# Plot record high and low temperatures
plt.plot(record_high, c="r", alpha=0.8, label="Record High (2005-2014)")
plt.plot(record_low, c="b", alpha=0.8, label="Record Low (2005-2014)")

# Scatter plot for 2015 record-breaking points
plt.scatter(broken_lows, record_low_2015.loc[broken_lows], s=20, c='black', label="2015 Record Low Break")
plt.scatter(broken_highs, record_high_2015.loc[broken_highs], s=20, c='b', alpha=0.8, label="2015 Record High Break")

# Legend & Titles
plt.legend()
plt.title("2015's Temperature Breaking Points Against 2005-2014 in Ann Arbor, Michigan, US")

# Fill the area between record high and low temperatures
plt.fill_between(record_high.index, record_low, record_high, facecolor='green', alpha=0.11)

# Aligning the plot
plt.gca().set_xlim([-1, 365])
plt.gca().set_ylim([-400, 450])

# Hide top and right plot spines
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Adjust x-axis ticks (Months)
month_ticks = [0, 30, 60, 90, 120, 150, 180, 210, 240, 270, 300, 330]
divs = [i+15 for i in month_ticks]
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.xticks(divs, month_names)

# Adjust y-axis tick labels
temp_labels = [f"{tick/10} °C" for tick in plt.gca().get_yticks()]  # Convert to degrees Celsius
plt.gca().set_yticks(plt.gca().get_yticks())
plt.gca().set_yticklabels(temp_labels)

# Save the figure
plt.savefig("Temp_Plot.png")

# Show the plot
plt.show()
```
