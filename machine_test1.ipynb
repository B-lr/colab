```python
#Converting the "Date" column to datetime

temp1['Date'] = pd.to_datetime(temp1['Date'])
temp1['Year'] = temp1['Date'].dt.year
year_counts = temp1['Year'].value_counts().sort_index()

print(year_counts)
```

```python
# Filtering data to 2005-2014
temp2 = temp1[(temp1["Date"] >= "2005-01-01") & (temp1["Date"] <= "2014-12-31")]
# Getting data during the year 2015
temp_2015 = temp1[(temp1["Date"] >= "2015-01-01") & (temp1["Date"] <= "2015-12-31")]
```

```python
# Removing Leap days
temp2 = temp2[temp2['Date'].dt.strftime('%m-%d') != '02-29'].copy()
```

```python
#sorting the temperature
temp2 = temp2.sort_values("Date")
```

```python
temp2.head()
```

```python
# Diving the data into two dataframes for high and low
high = temp2[temp2["Element"] == "TMAX"]
low = temp2[temp2["Element"] == "TMIN"]
```

```python
high.describe()
```

```python
#setting copies to avoid warnining
high = high.copy()
low = low.copy()

# Extract day-of-year after setting the year to 2015
high['dayofyear'] = high['Date'].apply(lambda x: x.replace(year=2015).timetuple().tm_yday)
low['dayofyear'] = low['Date'].apply(lambda x: x.replace(year=2015).timetuple().tm_yday)

# Group by day-of-year and get min/max
record_high = high.groupby("dayofyear")["Data_Value"].max()
record_low = low.groupby("dayofyear")["Data_Value"].min()
```

```python
#data 2015
temp_2015 = temp_2015.sort_values("Date")
temp_2015["Date"] = pd.to_datetime(temp_2015["Date"])

#splitting 2015 data to high and low
high_2015 = temp_2015[temp_2015["Element"] == "TMAX"]
low_2015 = temp_2015[temp_2015["Element"] == "TMIN"]


#record high and low of 2015

record_high_2015 = high_2015.copy()
record_high_2015["dayofyear"] = record_high_2015["Date"].dt.dayofyear
record_high_2015 = record_high_2015.groupby("dayofyear").max()


record_low_2015 = low_2015.copy()
record_low_2015["dayofyear"] = record_low_2015["Date"].dt.dayofyear
record_low_2015 = record_low_2015.groupby("dayofyear").min()
```

```python
# Reseting dataframes indexes
record_low = record_low.reset_index()
record_high = record_high.reset_index()
record_low_2015 = record_low_2015.reset_index()
record_high_2015 = record_high_2015.reset_index()
```

```python


# Identifying record-breaking days
broken_lows = record_low_2015[record_low_2015["Data_Value"] < record_low["Data_Value"]].index.tolist()
broken_highs = record_high_2015[record_high_2015["Data_Value"] > record_high["Data_Value"]].index.tolist()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(20,7))

# Plot record high and low temperatures
plt.plot(record_high["Data_Value"], c="r", alpha=0.8, label='Record High 2005-2014')
plt.plot(record_low["Data_Value"], c="b", alpha=0.8, label='Record Low 2005-2014')

# Scatter plot for 2015 temperature records
plt.scatter(broken_lows, record_low_2015.loc[broken_lows, 'Data_Value'], s=20, c='black', label='Record Low broken in 2015')
plt.scatter(broken_highs, record_high_2015.loc[broken_highs, 'Data_Value'], s=20, c='b', alpha=0.8, label='Record High broken in 2015')

plt.legend()
plt.title("2015's Temperature Breaking Points Against 2005-2014 in Ann Arbor, Michigan, US")

# Fill the area between record high and low temperatures
plt.fill_between(record_high.index, record_low["Data_Value"], record_high["Data_Value"], facecolor='yellow', alpha=0.11)

# Aligning the plot
plt.gca().axis([-1, 365, -400, 450])

# Hiding top and right plot spines
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

# Adjusting x-axis ticks (months)
month_ticks = [0, 30, 60, 90, 120, 150, 180, 210, 240, 270, 300, 330]
divs = [i+15 for i in month_ticks]
month_names = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.xticks(divs, month_names)

# Adjusting y-axis tick labels
temp = [f"{tick/10} °C" for tick in plt.gca().get_yticks()] # converting temp to actual degree C
plt.gca().set_yticks(plt.gca().get_yticks())
plt.gca().set_yticklabels(temp)

# Save the figure
plt.savefig('Temp_Plot.png')

# Show the plot
plt.show()

```
